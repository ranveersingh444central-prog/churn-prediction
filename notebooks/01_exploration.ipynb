# Telco Customer Churn - Step 1: Exploratory Data Analysis (EDA)
Dataset: `data/WA_Fn-UseC_-Telco-Customer-Churn.csv` (BlastChar on Kaggle, ~7,043 rows)

```python
from pathlib import Path
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

csv_path = Path("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
if not csv_path.exists():
    csv_path = Path("data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df = pd.read_csv(csv_path)
print(f"Shape: {df.shape}")
```

## 3. Missing Value Audit

```python
print("Standard null count:")
print(df.isna().sum())
```

## 4. Basic Statistics (Churn Rate & Key Categoricals)

```python
print("Churn Distribution:")
print(df["Churn"].value_counts(normalize=True) * 100)

for col in ["Contract", "InternetService", "PaymentMethod"]:
    print(f"\n{col} Value Counts:")
    print(df[col].value_counts())
    print(df[col].value_counts(normalize=True) * 100)
```
